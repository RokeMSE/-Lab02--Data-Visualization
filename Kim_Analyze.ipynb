## **Death_rate**

```python
df = pd.read_csv('Data/Death_rate/processed_death_rate.csv')

# Loại bỏ phần "[YRYYYY]" để chỉ giữ lại số năm
df['Year'] = df['Year'].str.extract('(\d+)').astype(int)

# Lọc dữ liệu từ năm 1974 đến 2022
df = df[(df['Year'] >= 1974) & (df['Year'] <= 2022)]

# Lọc các quốc gia cụ thể (ví dụ: các quốc gia trong khu vực Đông Nam Á)
countries_to_plot = ['Viet Nam', 'Cambodia', 'China', 'Thailand', 'Malaysia', 'Indonesia', 'Philippines']
df_filtered = df[df['Country Name'].isin(countries_to_plot)]

# Vẽ biểu đồ đường
plt.figure(figsize=(15, 10))
for country in countries_to_plot:
    country_data = df_filtered[df_filtered['Country Name'] == country]
    plt.plot(country_data['Year'], country_data['Death Rate'], label=country)

# Thêm tiêu đề và nhãn
plt.title('Tỷ lệ tử vong giữa các quốc gia Đông Nam Á từ năm 1974 đến năm 2022', fontsize=16)
plt.xlabel('Năm', fontsize=14)
plt.ylabel('Tỷ lệ tử vong (/ 1,000 people)', fontsize=14)
plt.legend(title='Quốc gia', fontsize=12)
plt.grid(True)
plt.show()
```

## **GDP**


```python
# Đọc dữ liệu
df_death = pd.read_csv('Data/Death_rate/processed_death_rate.csv')
df_gdp = pd.read_csv('Data/GDP/processed_gdp.csv')

# Đổi tên cột GDP thành Value 
df_gdp.rename(columns={'GDP': 'Value'}, inplace=True)

# Xử lý cột 'Year' để lấy đúng năm
df_death['Year'] = df_death['Year'].astype(str).str.extract(r'(\d+)').astype(int)
df_gdp['Year'] = df_gdp['Year'].astype(str).str.extract(r'(\d+)').astype(int)

# Lọc dữ liệu GDP chỉ lấy "GDP growth (annual %)"
df_gdp = df_gdp[df_gdp['Series Name'] == 'GDP growth (annual %)']
df_gdp = df_gdp[['Country Name', 'Year', 'Value']] 
df_gdp.rename(columns={'Value': 'GDP Growth (%)'}, inplace=True)

# Lọc khoảng năm 1990 - 2022
df_death = df_death[(df_death['Year'] >= 1990) & (df_death['Year'] <= 2022)]
df_gdp = df_gdp[(df_gdp['Year'] >= 1990) & (df_gdp['Year'] <= 2022)]

# Chọn các quốc gia Đông Nam Á
countries_to_plot = ['Viet Nam', 'Cambodia', 'China', 'Thailand', 'Malaysia', 'Indonesia', 'Philippines']
df_death_filtered = df_death[df_death['Country Name'].isin(countries_to_plot)]
df_gdp_filtered = df_gdp[df_gdp['Country Name'].isin(countries_to_plot)]

# Gộp dữ liệu theo quốc gia và năm
df_merged = pd.merge(df_death_filtered, df_gdp_filtered, on=['Country Name', 'Year'])

# Vẽ biểu đồ
plt.figure(figsize=(12, 6))
for country in countries_to_plot:
    country_data = df_merged[df_merged['Country Name'] == country]
    plt.plot(country_data['Year'], country_data['GDP Growth (%)'], label=f'{country} - GDP Growth')
    plt.plot(country_data['Year'], country_data['Death Rate'], label=f'{country} - Death Rate', linestyle='dashed')

plt.xlabel('Năm', fontsize=12)
plt.ylabel('Tỷ lệ tử vong / Tăng trưởng GDP', fontsize=12)
plt.title('Tỷ lệ tử vong và tăng trưởng GDP ở Đông Nam Á (1990 - 2022)', fontsize=14)
plt.legend(title='Quốc gia', fontsize=10)
plt.grid(True)
plt.show()
```

### **Phân tích biểu đồ: Tỷ lệ tử vong và tăng trưởng GDP ở Đông Nam Á (1990 - 2022)**  

Biểu đồ hiển thị hai chỉ số quan trọng theo thời gian cho các quốc gia Đông Nam Á:  
- **Tăng trưởng GDP (GDP Growth %)** – đường liền nét.  
- **Tỷ lệ tử vong (Death Rate)** – đường đứt nét.  

#### **1. Xu hướng chung**  
**Tăng trưởng GDP (%) có biến động mạnh hơn so với tỷ lệ tử vong**  
   - GDP có xu hướng tăng trưởng dương trong phần lớn giai đoạn, với một số cú sốc kinh tế rõ ràng (1997-1998, 2008-2009, 2020).  
   - Các cú sốc kinh tế ảnh hưởng đến tất cả các quốc gia với mức giảm GDP mạnh.  

**Tỷ lệ tử vong có xu hướng ổn định và giảm nhẹ**  
   - Đường đứt nét của tỷ lệ tử vong không có nhiều biến động lớn.  
   - Xu hướng chung là tỷ lệ tử vong giảm dần hoặc giữ ổn định, phản ánh sự phát triển của y tế, dinh dưỡng và điều kiện sống.  


#### **2. Những giai đoạn đáng chú ý**  
**Khủng hoảng tài chính châu Á (1997-1998)**  
   - GDP của các nước như Thái Lan, Indonesia, Malaysia giảm mạnh.  
   - Việt Nam và Trung Quốc ít bị ảnh hưởng hơn.  

**Khủng hoảng kinh tế toàn cầu (2008-2009)**  
   - GDP giảm sút nhưng không nghiêm trọng như 1997-1998.  

**Đại dịch COVID-19 (2020-2021)**  
   - GDP của hầu hết các nước đều giảm mạnh.  
   - Cambodia chịu tác động nghiêm trọng nhất, thể hiện qua đường GDP đi xuống sâu.  


#### **3. So sánh theo quốc gia**  
- **Việt Nam**: GDP tăng trưởng khá ổn định, ít chịu tác động tiêu cực lớn.  
- **Trung Quốc**: Duy trì mức tăng trưởng cao trong thời gian dài, ít cú sốc.  
- **Indonesia, Malaysia, Thái Lan**: Có mức tăng trưởng khá tốt nhưng chịu ảnh hưởng mạnh từ các cuộc khủng hoảng.  
- **Cambodia**: GDP biến động lớn nhất, đặc biệt giảm sâu vào những năm khủng hoảng.  
- **Philippines**: GDP biến động nhẹ nhưng có xu hướng tăng đều.  


### **Kết luận**  
- **Tăng trưởng GDP và tỷ lệ tử vong có mối quan hệ ngược chiều**: Khi nền kinh tế phát triển, điều kiện sống cải thiện, dẫn đến tỷ lệ tử vong giảm dần.
