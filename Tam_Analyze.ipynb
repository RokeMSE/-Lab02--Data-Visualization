#### **GDP**

```python
print(gdp.isnull().sum())
```

```python
# Chỉ lấy các cột có chứa dữ liệu theo năm
year_cols = [col for col in gdp.columns if "[YR" in col]

# Nội suy tuyến tính trên các cột này
gdp[year_cols] = gdp[year_cols].interpolate(method="linear", axis=1)
```

```python
# Điền dữ liệu còn thiếu từ trước
gdp[year_cols] = gdp[year_cols].ffill(axis=1)
gdp[year_cols] = gdp[year_cols].bfill(axis=1)
```

```python
# Xóa các dòng có tất cả giá trị còn thiếu
gdp = gdp.dropna(subset=year_cols, how="all")
```

```python
print(gdp.isnull().sum())
```

```python
gdp.to_csv('processedByTam/processed_gdp.csv', index=False)
```

#### **Life_Expentancy**
